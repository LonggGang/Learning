Setup

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB 
from sklearn.metrics import accuracy_score

```

Load Data

```python
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
#df.head()
#df.info()

df2 = pd.DataFrame(data.target, columns=['target'])
#df2.head()
```

Split Data

```python
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.4, random_state=42)
```

Building and Training Model

```python
model1 = LogisticRegression()
model1.fit(X_train, y_train)
y_pred1 = model1.predict(X_test)

print(f'LogisticRegression: {accuracy_score(y_test, y_pred1) * 100}') 

model2 = GaussianNB()
model2.fit(X_train, y_train)
y_pred2 = model2.predict(X_test)

print(f'GaussianNB: {accuracy_score(y_test, y_pred2) * 100}')
```
